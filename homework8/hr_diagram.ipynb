```python
import numpy as np
import matplotlib.pyplot as plt
```

## 3.1 Importing Data with NumPy

```python
data = np.loadtxt("/Users/andiiianna/python_decal_fa26/course_assignments/homework7/star_data.txt", skiprows = 1)
```

## 3.2 Analyzing Importes Data

```python
data.shape
# there are 2642 row and 2 columns 

magnitude = data[:,0]
tempurature = data[:,1]

print(f"The minimum value of the magnitudes is {np.min(magnitude)} and it corresponds to star {magnitude.min()}")
print(f"The maximum value of the magnitudes is {np.max(magnitude)} and it corresponds to star {magnitude.max()}")
print(f"The minimum value of the tempuratures is {np.min(tempurature)} and it corresponds to star {tempurature.min()}")
print(f"The maximum value of the tempuratures is {np.max(tempurature)} and it corresponds to star {tempurature.max()}")

# star 2.24 is brighter and star 12.49 is dimmer 
```

## 3.3 Plot and HR Diagram

```python
log_magnitude = np.log(magnitude)
log_tempurature = np.log(tempurature)

plt.figure(figsize = (8,6))

plt.scatter(log_tempurature, log_magnitude, label = "Stars", c = tempurature, cmap = "coolwarm")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

brightest = np.argmin(magnitude)
dimmest = np.argmax(magnitude)
hottest = np.argmax(tempurature)
coldest = np.argmin(tempurature)
plt.scatter(log_tempurature[brightest], log_magnitude[brightest], color = "Black", marker = "*", s = 100, label = "Brightest")
plt.scatter(log_tempurature[dimmest], log_magnitude[dimmest], color = "purple", marker = "*", s = 100, label = "Dimmest")
plt.scatter(log_tempurature[hottest], log_magnitude[hottest], color = "darkgreen", marker = "*", s = 100, label = "Hottest")
plt.scatter(log_tempurature[coldest], log_magnitude[coldest], color = "orange", marker = "*", s = 100, label = "Coldest")

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Tempurature(logK)")
plt.ylabel("Magnitude")


plt.legend()
plt.show()
```
