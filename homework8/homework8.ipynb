```python
import numpy as np 
import matplotlib.pyplot as plt
```

## 2 Matplotlib

```python
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y

x = np.linspace(0, 2*np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

fig, axes = plt.subplots(1, 3, figsize = (25,5))

for A, w in zip(amplitude,frequency):
    axes[0].plot(x, make_sine_wave(x, A, 1), label = f"A = {A}")
    axes[1].plot(x, make_sine_wave(x, 1, w), label = f"w = {w}")
    axes[2].plot(x, make_sine_wave(x, A, w), label = f"A = {A}, w = {w}")

axes[0].set_title("Varying Amplitude", fontsize = 30)
axes[0].set_xlabel("x", fontsize = 15)
axes[0].set_ylabel("y", fontsize = 15)
axes[0].legend()

axes[1].set_title("Varying Frequency", fontsize = 30)
axes[1].set_xlabel("x", fontsize = 15)
axes[1].set_ylabel("y", fontsize = 15)
axes[1].legend()

axes[2].set_title("Varying Amplitude and Frequency", fontsize = 30)
axes[2].set_xlabel("x", fontsize = 15)
axes[2].set_ylabel("y", fontsize = 15)
axes[2].legend(fontsize = 7)

plt.tight_layout()
plt.show()
```
