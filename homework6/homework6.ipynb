## 4 Prime Star System 

```python
arr = np.array([[2, 3, 5],
               [14, 6, 8],
               [11, 13, 17],
               [7, 10, 13]])
def at_least_one_prime(arr):
    result = []
    for row in arr: 
        for num in row:
            if num > 1:
                prime = True 

                for i in range(2,num):
                    if num % i == 0:
                        prime = False
                if prime:
                    result.append(row)
                    break
    return np.array(result)
print(at_least_one_prime(arr))
```

## 5 Let's Play Checkers

```python
board = np.zeros((8,8))

def checkerboard(board):
    for i in range(8):
        if i % 2 ==0:
            board[i, ::2] =1
        else:
            board[i, 1::2] = 1
    return board 
print(checkerboard(board))
```

## 6 The Expanding Universe 

```python
universe = np.array(['galaxy', 'clusters'])
def expansion(arr, n):
    result = []
    for word in arr:
        new_word = ""

        for letter in word: 
            new_word = new_word + letter + " " * n
        result.append(new_word)

    return np.array(result)

print(expansion(universe,1))
    
```
