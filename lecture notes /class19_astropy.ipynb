### fits 

```python
from astropy.io import fits
import matplotlib.pyplot as plt
```

```python
with fits.open("astropy_data.fits") as hdul:
    header = hdul[0].header
    data = hdul[0].data
```

```python
plt.imshow(data, cmap = 'plasma')
plt.colorbar()
plt.show()
```

```python
from matplotlib.colors import LogNorm
plt.imshow(data, cmap ="gray", norm = LogNorm())
plt.colorbar()
plt.show()
```

```python
from matplotlib.colors import LogNorm
plt.imshow(data, cmap ="inferno", norm = LogNorm())
plt.colorbar()
plt.show()
```
