```python
import numpy as np
import matplotlib.pyplot as plt

# Given values
d = 4.6692016

n = np.array([1, 2, 3, 4])
g = np.array([1.0663, 1.0793, 1.0821, 1.0827])

# Calculate d^(-n)
x = d**(-n)

# Find best-fit line
slope, intercept = np.polyfit(x, g, 1)

# Plot data points
plt.scatter(x, g, label="Given data")

# Plot best-fit line
plt.plot(x, slope*x + intercept, label="Best-fit line")

# Labels and title
plt.xlabel("d^(-n)")
plt.ylabel("g_n")
plt.title("Feigenbaum Thresholds")

plt.legend()
plt.grid()
plt.show()

# Print vertical intercept
print("Estimated gc =", intercept)
```
